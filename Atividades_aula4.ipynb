```python
#atividade1

número = int (input ("Qual o número?"))

if número % 2 == 0 or número % 3 == 0:
  print ("O número {} é primo " .format(número))
else:
  print ("O número {} não é primo " .format(número))



#Atividade2

import math

n = float (input ("Qual o número?"))
b = float (input ("Qual o valor de b?"))

while True:
  p = (b+ (n/b) ) / 2
  quadrado_p = p ** 2
  if abs(n - quadrado_p) < 0.0001:
    break
  b = p

raiz_oficial = math.sqrt(n)

print(f"\nRaiz do método Newton: {p:.6f}")
print(f"Raiz da função SQRT(n):    {raiz_oficial:.6f}")
```
