```python
#Atividade número I

import random

tentativas = []
numero = random.randint(0, 15)

while True:
  tentativa = int (input ("Qual sua tentativa? "))
  if numero != tentativa:
    print ("_______________________________")
    print ("Tente novamente ")
    tentativas.append(tentativa)
  else:
    break
print ("_______________________________")
print ("Você acertou o número", numero)
print ("Você tentou", len(tentativas), "vezes")
```

```python
#Atividade número II
import csv

produto = []
preço = []
quantidade = []

while True:
  print ("1-Comprar")
  print ("2-Sair")
  opação = int (input ("Oque você deseja?"))
  if opação == 1:
    compras = input ("Qual produto você deseja comprar? ")
    produto.append(compras)
    quantidades = input ("Qual a quantidade que deseja desse produto?")
    quantidade.append(quantidades)
    preço_produto = int (input ("Qual o preço desse produto?"))
    preço.append(preço_produto)
  else:
    break
v = len(produto)


with open("dados.csv", "w", newline="", encoding="utf-8") as arquivo:

  leitor = csv.writer(arquivo)

  leitor.writerow(["Quantidade", "Produto", "Preço"])
  for i in range(v):
    leitor.writerow([quantidade[i], produto[i], preço[i]])

```
