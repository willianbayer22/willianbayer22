```python
import random
import json
import csv


lista = []
for i in range(20):
  num_aleatorio = random.randint(1, 999)
  lista.append(num_aleatorio)
print ("Valor original" , lista)


listax2 = []
multiplicação = []
for numero in lista:
  listax2.append(numero)
  soma = sum(listax2)
  resultado = soma * 2
  multiplicação.append(resultado)
  resultado = None
  listax2.pop()
print ("Valor *2" , multiplicação)


listax10 = []
multiplicaçãox10 = []
for numero in lista:
  listax10.append(numero)
  soma2 = sum(listax10)
  resultado2 = soma2 * 10
  multiplicaçãox10.append(resultado2)
  resultado2 = None
  listax10.pop()
print ("Valor *10" , multiplicaçãox10)

```

```python
with open("dados.txt", "w", encoding="utf-8") as arquivo:

    arquivo.write("Valor original | Valor *2 | Valor *10\n")

    for i in range(20):

        arquivo.write(
            f"{lista[i]} | {multiplicação[i]} | {multiplicaçãox10[i]}\n"
        )
```

```python
dados = []

for i in range(20):

    dados.append({
        "valor original": lista[i],
        "valor *2": multiplicação[i],
        "valor *10": multiplicaçãox10[i]
    })


with open("dados.json", "w", encoding="utf-8") as arquivo:

    json.dump(dados, arquivo, indent=4)

```

```python
with open("dados.csv", "w", newline="", encoding="utf-8") as arquivo:

    escritor = csv.writer(arquivo)

    escritor.writerow(["valor original", "valor *2", "valor *10"])

    for i in range(20):

        escritor.writerow([
            lista[i],
            multiplicação[i],
            multiplicaçãox10[i]
        ])


print("Dados salvos em TXT, JSON e CSV!")
```
