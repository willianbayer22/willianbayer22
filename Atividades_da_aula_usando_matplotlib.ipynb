```python
# Importação das bibliotecas pandas, numpy e matplotlib.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}
# Criação do dataFrame
df = pd.DataFrame(dados)

print(df) # Print de teste
```

```python
# 1. Exibindo as primeiras linhas, as dimensões do DataFrame e as estatísticas descritivas.
df.head()
```

```python
# 2. A média, o maior valor e o menor valor de vendas de cada produto.

# Eliminando a coluna "Mês" e deixando apenas as númericas
vendas = df[["Notebook", "Mouse", "Teclado"]].values

# Calculando as estatísticas usando NumPy ao longo do eixo 0 (linhas)
medias = np.mean(vendas, axis=0)
maiores = np.max(vendas, axis=0)
menores = np.min(vendas, axis=0)

# Exibindo os resultados formatados
produtos = ["Notebook", "Mouse", "Teclado"]

print("Estatísticas de Vendas por Produto:")
for i, produto in enumerate(produtos):
    print(f"\n--- {produto} ---")
    print(f"Média: {medias[i]:.2f}")
    print(f"Maior Valor: {maiores[i]}")
    print(f"Menor Valor: {menores[i]}")
```

```python
# 3. Grafico de linha comparando as vendas dos três produtos ao longo dos meses.
# Tamanho do grafico
plt.figure(figsize=(10, 6))

# Passar os valores
meses = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio", "Junho"]
Not = [20, 25, 22, 30, 35, 40]
Mouse = [50, 55, 48, 60, 65, 70]
Teclado = [30, 28, 35, 40, 38, 45]

# Plota a linha para cada produto utilizando a coluna 'Mes' no eixo X
plt.plot(meses, Not, label="Notebook")
plt.plot(meses, Mouse, label="Mouse")
plt.plot(meses, Teclado, label="Teclado")

# Legenda com as cores de cada produto
plt.legend()

# Adiciona linhas de grade ao fundo para facilitar a leitura
plt.grid(True, linestyle="--", alpha=0.6)

# Mostra o grafico na tela
plt.show()
```

```python
# 4. Gráfico de barras com o total vendido por produto no período.
# Tamanho do grafico
plt.figure(figsize=(10, 7))

# Criar uma variavel com os nomes
Produtos = "Notebook", "Mouse", "Teclado"

# Passar os valores
Not = [20, 25, 22, 30, 35, 40]
Mouse = [50, 55, 48, 60, 65, 70]
Teclado = [30, 28, 35, 40, 38, 45]

# Fazer a soma de todos os valores
Tnot = sum(Not)
Tmouse = sum(Mouse)
Tteclado = sum(Teclado)

# Criar uma variavel com os valores
Valores = Tnot, Tmouse, Tteclado

# Criar o grafico
plt.bar(Produtos, Valores, color=["red", "blue", "green"])

# Adiciona linhas de grade ao fundo para facilitar a leitura
plt.grid(axis="y", linestyle="--", alpha=0.6)

# Mostra o grafico na tela
plt.show()
```

```python
# 5. Histograma com os valores mensais de vendas do produto mouse.
# Tamanho do grafico
plt.figure(figsize=(10, 7))

# Passar os valores
# Criar uma variavel com os meses e outra com os valores
Meses = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio", "Junho"]
Mouse = [50, 55, 48, 60, 65, 70]

# Criar o grafico
plt.bar(Meses, Mouse)

# Adiciona linhas de grade ao fundo para facilitar a leitura
plt.grid(axis="y", linestyle="--", alpha=0.6)

# Mostra o grafico na tela
plt.show()
```

6. Interpretação: Identifique o produto com maior volume total, o mês de maior venda de cada produto e uma conclusão baseada nos gráficos.

O produto com o maior volume foi o mouse.
O mês que vai vendeu "Notebook, mouse e teclado" foi o mês de junho.
Ao analisar qualquer um dos gráfico, podemos concluir que há uma tendência clara de crescimento contínuo nas vendas ao longo do semestre.

```python
# Atividade extra, calculando a média de venda dos mouse
# Variaveis com os valores
Meses = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"]
Mouse = [50, 55, 48, 60, 65, 70]

# Calcular a média semestral do Mouse
soma = sum(Mouse)
quantidade = len(Mouse)
media_mouse = soma/quantidade  # O resultado é 58.0

# Tamanho do grafico
plt.figure(figsize=(10, 7))

# Plotar as barras com as cores condicionais
barras = plt.bar(Meses, Mouse)

# Adicionar a linha de corte (Média Semestral)
plt.axhline(y=media_mouse, color="darkred", linestyle="--", linewidth=2.5,
            label=f"Média Semestral ({media_mouse:.1f} un)")

# Mostrar grafico
plt.show()
```
