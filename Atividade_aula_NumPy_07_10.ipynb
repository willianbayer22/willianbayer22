```python
import numpy as np
```

```python
# Criando os dados
dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array(["Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"])

estrutura = ["Pedidos", "Faturamento", "Clientes","Ticket Médio", "Reclamações"]

#print (estrutura)
#print (dados)
#print (dias)

# Definindo a largura de cada coluna
larguras = [12, 12, 16, 12, 16, 14]

# Criando o cabeçalho centralizado
cabecalho = [
    "Dia", "Pedidos", "Faturamento",
    "Clientes", "Ticket Médio", "Reclamações"
]

print("".join(
    nome.center(largura)
    for nome, largura in zip(cabecalho, larguras)
))

# Linha separadora
print("-" * sum(larguras))

# Exibindo os dados centralizados
for i in range(len(dias)):
    valores = [
        dias[i],
        f"{dados[i, 0]:.0f}",
        f"{dados[i, 1]:.0f}",
        f"{dados[i, 2]:.0f}",
        f"{dados[i, 3]:.2f}",
        f"{dados[i, 4]:.0f}"
    ]

    print("".join(
        valor.center(largura)
        for valor, largura in zip(valores, larguras)
    ))
```

```python
# Parte 1 — Conhecendo os dados

# Quantidade de linhas e colunas
LC = dados.shape
print (" A quantidade de linhas e colunas é:", LC)

# Quantidade total dos valores
V = dados.size
print ("\n A quantidade de valores é:", V)

# Número de dimensões do array
D = dados.ndim
print ("\n O número de dimensões do array é:", D)

# Tipo dos dados armazenados
T = dados.dtype
print ("\n Tipo de dados armazenados são:", T)
```

```python
# Parte 2 — Acessando os dados

# Mostre os dados referentes à sexta-feira.
sexta = dados[4]
print("Dados de Sexta-feira:", sexta)

# Mostre somente o faturamento de todos os dias.
faturamento = dados[:, 1]
print("\nFaturamento de todos os dias:", faturamento)

#Mostre somente a quantidade de reclamações.
reclamações = dados[:, 4]
total = 0
i = 0
# Percorre a lista enquanto o índice for menor que reclamações
while i < len(reclamações):
  total += reclamações[i]
  i += 1
print("\nA quantidde total de reclamçãoes é:", total)

# Mostre os dados de quinta-feira até domingo
QD = dados[3:7]
print("\nDados de quinta até domingo:")
print (QD)

# Mostre somente as colunas: pedidos, faturamento, clientes.
colunas = dados[:, 0:3]
print ("\nPedidos Faturamento Clientes")
print (colunas)
```

```python
#Parte 3 — Operações numéricas

# Calcule o total de pedidos realizados durante a semana.
TP = dados [:, 0]
total = 0
i = 0
while i < len(TP):
  total += TP[i]
  i += 1
print("O total de pedidos durente a semana foi", total)

# Calcule o faturamento total
faturamento = dados [:, 1]
total2 = 0
i = 0
while i < len(faturamento):
  total2 += faturamento[i]
  i += 1
print ("\nO total do faturamento é", total2)

# Calcule a média diária de pedidos.
media = dados [:, 0]
total3 = 0
i = 0
while i < len(media):
  total3 += media[i]
  i += 1
  final = round(total3 / i, 2)
print ("\nA média diaria dos pedidos é:", final)

# Calcule a média diária de faturamento.
MD = dados [:, 1]
total4 = 0
i = 0
while i < len(MD):
  total4 += MD[i]
  i += 1
  MDfinal = round(total4 / i, 2)
print ("\nA média diaria do faturamento é:", MDfinal)

# Calcule a média de reclamações por dia
reclamações = dados [:, 4]
total5 = 0
i = 0
while i < len(reclamações):
  total5 += reclamações[i]
  i += 1
  Rfinal = round(total5 / i, 2)
print ("\nA média diaria de reclamações é:", Rfinal)
```

```python
# Parte 4 — Análise por dia

pedidos = dados [:, 0]
# Dia com o maior número de pedidos
Rmaior = dias[np.argmax(pedidos)]
print ("O dia com o maior número de pedidos foi:", Rmaior)

# Dia com o menor número de pedidos
Rmenor = dias[np.argmin(pedidos)]
print ("\nO dia com o menor número de pedidos foi:", Rmenor)

faturamento = dados [:, 1]
# Dia com maior faturamento
Fmaior = dias[np.argmax(faturamento)]
print ("\nDia com maior faturamento foi:", Fmaior)

# Dia com menor faturamento
Fmenor = dias[np.argmin(faturamento)]
print ("\nDia com o menor faturamento foi:", Fmenor)
```

```python
# Parte 5 — Máscaras e filtros

# Mostre os dias em que o faturamento foi superior a 12000
faturamento = dados[:, 1]
S12000 = dias[faturamento >= 12000]
print("Dias que o faturamento foi superior a 12000:", S12000)

# Mostre os dias em que foram realizados mais de 150 pedidos
pedidos = dados[:, 0]
P150 = dias[pedidos >= 150]
print ("\nDias em que foram realizados mais de 150 pedidos", P150)

# Mostre os dias que tiveram mais de 4 reclamações
reclamações = dados [:, 4]
R4 = dias[reclamações >= 4]
print ("\nDias que tiveram mais de 4 reclamações", R4)

# Mostre os faturamentos dos dias que tiveram mais de 150 pedidos
FD150 = faturamento[pedidos >= 150]
print ("\nFaturamento dos dias com mais de 150 pedidos:", FD150)

# Quantos dias tiveram faturamento abaixo da média semanal?
MD = dados [:, 1]
total4 = 0
i = 0
while i < len(MD):
  total4 += MD[i]
  i += 1
  MDfinal = round(total4 / i, 2)
#Print "17714"

Mfat = dias[faturamento >= 17714]
print("\nDias tiveram faturamento abaixo da média semanal:", Mfat)
```

```python
# Parte 6 — Classificação dos dias
# Classificando os dias de acordo com o faturamento
classificacao = np.where(
    faturamento >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

# Apresente a classificação correspondente a cada dia.
print("Classificação por dia:")
print(np.column_stack((dias, classificacao)))

# Quantos dias foram classificados como ?

quantidade_alto = np.count_nonzero(
    classificacao == "Alto desempenho"
)

quantidade_normal = np.count_nonzero(
    classificacao == "Desempenho normal"
)

print("\nDias de alto desempenho:", quantidade_alto)
print("Dias de desempenho normal:", quantidade_normal)
```

```python
# Parte 7 — Estatística
#Calculando as estatísticas do faturamento

# Média do faturamento
media = np.mean(faturamento)
# Mediana do faturamen

mediana = np.median(faturamento)
# Desvio padrão
desvio_padrao = np.std(faturamento)
# Maximo
minimo = np.min(faturamento)
# Minimo
maximo = np.max(faturamento)

print("Média: R$", round(media, 2))
print("\nMediana: R$", round(mediana, 2))
print("\nDesvio padrão: R$", round(desvio_padrao, 2))
print("\nMínimo: R$", minimo)
print("\nMáximo: R$", maximo)
```

```python
# Parte 8 — Trabalhando com axis

# A média de cada indicador durante a semana.
medias = np.mean(dados, axis=0)
print("Média de cada indicador:")
print(medias)

# O maior valor observado em cada indicador.
maiores = np.max(dados, axis=0)
print("\nMaior valor de cada indicador:")
print(maiores)

#  O menor valor observado em cada indicador.
menores = np.min(dados, axis=0)
print("\nMenor valor de cada indicador:")
print(menores)

# Calculando o desvio padrão de cada indicador
desvios = np.std(dados, axis=0)

# Coeficiente de variação em porcentagem
coeficientes = (desvios / medias) * 100

print("\nCoeficiente de variação (%):")
print(coeficientes)

# Identificando o indicador com maior variação relativa
indice_maior_variacao = np.argmax(coeficientes)

indicadores = np.array([
    "Pedidos",
    "Faturamento",
    "Clientes",
    "Ticket médio",
    "Reclamações"
])

print(
    "Indicador com maior variação relativa:",
    indicadores[indice_maior_variacao]
)
```
