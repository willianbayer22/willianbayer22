```python
nota = float (input("Digite o valor da nota"))
if nota < 60:
  print ("Sua nota {} foi insuficiente!" .format(nota))
elif nota >60 and nota <75:
  print ("Sua nota {} foi C!" .format(nota))
elif nota >75 and nota <90:
  print ("Sua nota {} foi B!" .format(nota))
else:
  print ("Sua nota {} foi sufuciente" .format(nota))

```

```python
def calcula_percento (valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print ("20% de 300 são: {:10.2f}" .format (calcula_percento (300,20)))
print ("100% de 50 são: {:10.2f}" .format (calcula_percento (50)))
```
