```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import re
import time

# ==============================
# CONFIGURAÇÃO DO CHROME
# ==============================

options = Options()
options.binary_location = "/usr/bin/google-chrome"

options.add_argument("--headless=new")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--disable-gpu")
options.add_argument("--window-size=1920,1080")
options.add_argument("--lang=pt-BR")

driver = webdriver.Chrome(options=options)

try:

    # ==============================
    # ENTRADA DA CIDADE
    # ==============================

    cidade = input("Digite o nome da cidade: ").strip()

    if cidade == "":
        raise Exception("Cidade não informada.")

    print("\nPesquisando previsão para:", cidade)

    # ==============================
    # TESTE COM SANTA ROSA
    # ==============================

    if cidade.lower() == "santa rosa":

        url = "https://www.climatempo.com.br/previsao-do-tempo/cidade/781/santa-rosa-rs"

    else:

        raise Exception(
            "Por enquanto estamos testando com Santa Rosa."
        )

    print("\nAbrindo:")
    print(url)

    # ==============================
    # ABRIR PÁGINA
    # ==============================

    driver.get(url)

    time.sleep(5)

    # ==============================
    # PEGAR TEXTO DA PÁGINA
    # ==============================

    texto = driver.find_element(
        By.TAG_NAME,
        "body"
    ).text

    # Mostrar uma parte do texto para verificar
    print("\nPágina carregada!")

    # ==============================
    # PROCURAR TEMPERATURAS
    # ==============================

    temperatura_minima = None
    temperatura_maxima = None

    # Procura padrão:
    # 17° 22°

    padrao = re.search(
        r'(\d{1,2})°\s*(\d{1,2})°',
        texto
    )

    if padrao:

        temperatura_minima = padrao.group(1)
        temperatura_maxima = padrao.group(2)

    # ==============================
    # MOSTRAR RESULTADO
    # ==============================

    print("\n")
    print("=" * 50)
    print("          PREVISÃO DO TEMPO")
    print("=" * 50)

    print("Cidade:", cidade)

    if temperatura_minima:
        print(
            "Temperatura mínima:",
            temperatura_minima,
            "°C"
        )
    else:
        print(
            "Temperatura mínima: não encontrada"
        )

    if temperatura_maxima:
        print(
            "Temperatura máxima:",
            temperatura_maxima,
            "°C"
        )
    else:
        print(
            "Temperatura máxima: não encontrada"
        )

    print("=" * 50)

except Exception as erro:

    print("\nOcorreu um erro:")
    print(erro)

finally:

    driver.quit()
```
