```python
a = int (input ("Qual o primeiro valor que deseja?"))
o = (input ("Qual operação deseja (+, -, /, *, raiz): "))
if o != "raiz":
  b = int (input ("Qual o segundo valor que deseja?"))
else:

  class Calculadora:
    def __init__(self, a: int, b: int):
        self.a = a
        self.b = b

    def somar(self):
        r = self.a + self.b
        print(a, o, b, "=", r)

    def subtrair(self):
        s = self.a - self.b
        print(a, o, b, "=", s)

    def dividir(self):
        d =  self.a / self.b
        print(a, o, b, "=", d)

    def multiplicar(self):
        m =  self.a * self.b
        print(a, o, b, "=", m)

    def raiz_quadrada(self):
        if self.a < 0:
            print("Número invalido")
            return

        palpite = self.a / 2.0 if self.a > 0 else 0.0
        while abs(palpite * palpite - self.a) > 1e-5:
            palpite = (palpite + self.a / palpite) / 2.0

        print(f"A raiz quadrada de {self.a} é aproximadamente: {palpite:.4f}")

minha_calculadora = Calculadora(a, b)

if o == "-":
    minha_calculadora.subtrair()
elif o == "+":
    minha_calculadora.somar()
elif o == "/":
    minha_calculadora.dividir()
elif o == "*":
    minha_calculadora.multiplicar()
elif o.lower() == "raiz":
    minha_calculadora.raiz_quadrada()
else:
    print("Essa operação não existe")
```
