```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
```

```python
#localizar a linha que esta minha cidade.
cidade = df[df["nome_municipio"] == "Senador Salgado Filho"]

(cidade)
```

```python
#Verificar em qual mes teve a menor temperatura
temperaturas = df.iloc[5031][[
    "temp_media_jan", "temp_media_fev", "temp_media_mar",
    "temp_media_abr", "temp_media_mai", "temp_media_jun",
    "temp_media_jul", "temp_media_ago", "temp_media_set",
    "temp_media_out", "temp_media_nov", "temp_media_dez"
]]
mes = temperaturas.idxmin()
valor = temperaturas.min()

print("Mês com menor temperatura:", mes.replace("temp_media_", ""))
print("Menor temperatura:", valor)
```

```python
#Verificar qual o mês de maior déficit hídrico
deficit = df.iloc[5031][[
    "def_jan", "def_fev", "def_mar",
    "def_abr", "def_mai", "def_jun",
    "def_jul", "def_ago", "def_set",
    "def_out", "def_nov", "def_dez"
]]

deficit.idxmax()

print("Mês:", deficit.idxmax())
print("Maior déficit:", deficit.max())
```

```python
#Separar os mês por estação
temperaturas = df.iloc[5031]

verao = [
    temperaturas["temp_media_dez"],
    temperaturas["temp_media_jan"],
    temperaturas["temp_media_fev"]
]

outono = [
    temperaturas["temp_media_mar"],
    temperaturas["temp_media_abr"],
    temperaturas["temp_media_mai"]
]

inverno = [
    temperaturas["temp_media_jun"],
    temperaturas["temp_media_jul"],
    temperaturas["temp_media_ago"]
]

primavera = [
    temperaturas["temp_media_set"],
    temperaturas["temp_media_out"],
    temperaturas["temp_media_nov"]
]


#Calcular a média de cada estação
media_verao = sum(verao) / len(verao)
media_outono = sum(outono) / len(outono)
media_inverno = sum(inverno) / len(inverno)
media_primavera = sum(primavera) / len(primavera)

#Print pra ver os resultados
print("Média do Verão:", media_verao)
print("Média do Outono:", media_outono)
print("Média do Inverno:", media_inverno)
print("Média da Primavera:", media_primavera)
```
