```python
# Desafio 1 – Análise de notas de uma turma

lista = []

estudantes = int (input ("Qual o número de estudantes?"))

while True:
  if estudantes != len(lista) :
    notas = int (input ("Qual são as notas dos estudantes?"))
    lista.append(notas)
  else:
    break

print ("A maior nota é: ", max (lista))
print ("A menor nota é: ", min (lista))

soma = sum(lista)
quantidade = len(lista)
media = soma / quantidade
print ("A média da turma é: ", media)

aprovados = []
reprovados = []
for lista in lista:
  if lista >= 6:
    aprovados.append(lista)
  else:
    reprovados.append(lista)

print ("A quantidade de alunos aprovados é: ", len(aprovados))
print ("A quantidade de alunos reporvados é: ", len(reprovados))
```

```python
# Desafio 3 – Sistema de Estatísticas

lista = []
while True:
  print ("Precione '0' para encerrar")
  valores = int (input ("Qual o valor deseja?"))
  if valores != 0:
    lista.append(valores)
  else:
    break

soma = sum(lista)
print ("A soma total da lista é: ", soma)

quantidade = len(lista)
media = soma / quantidade
print ("A média da lista é: ", media )
#___________________________

lista.sort()
meio = quantidade // 2
if quantidade % 2 == 1:
    mediana = lista[meio]
else:
    mediana = (lista[meio - 1] + lista[meio]) / 2
print("A mediana da lista é:", mediana)

#___________________________
print ("O maior valor da lista é: ", max (lista))
print ("O menor valor da lista é: ", min (lista))

pares = []
impares = []
for lista in lista:
  if lista % 2 == 0:
        pares.append(lista)
  else:
        impares.append(lista)

print("Pares:", pares)
print("Ímpares:", impares)
```
